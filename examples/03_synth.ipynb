# 03 SynthWidget Tutorial

Use multiple oscillator types and compare waveform behavior.

```python
import ipywidgets as widgets

from nbplay import SynthWidget

lead = SynthWidget(oscillator_type="saw", frequency=660.0, amplitude=0.55)
bass = SynthWidget(oscillator_type="square", frequency=110.0, amplitude=0.65)
texture = SynthWidget(oscillator_type="noise", amplitude=0.18)

widgets.HBox([lead, bass, texture])
```

Beyond one oscillator: a second oscillator a few cents sharp thickens the sound, a low-pass filter with its own envelope gives each note a bright attack that closes down, and an LFO adds vibrato. Notes from a sequencer, a clip or a launched slot on the synth's track play through the whole patch.

```python
pluck = SynthWidget(
    oscillator_type="saw",
    frequency=220.0,
    amplitude=0.5,
    osc2_type="saw",
    osc2_level=0.8,
    osc2_detune=9.0,
    cutoff=600.0,
    resonance=4.0,
    filter_env=3.0,
    filter_decay=0.3,
    filter_sustain=0.1,
    decay=0.4,
    sustain=0.3,
    lfo_target="pitch",
    lfo_depth=0.1,
)
pluck
```
