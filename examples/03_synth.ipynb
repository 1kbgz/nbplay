# 03 SynthWidget Tutorial

Use multiple oscillator types and compare waveform behavior.

```python
import ipywidgets as widgets

from nbplay import SynthWidget

lead = SynthWidget(oscillator_type="saw", frequency=660.0, amplitude=0.55)
bass = SynthWidget(oscillator_type="square", frequency=110.0, amplitude=0.65)
texture = SynthWidget(oscillator_type="noise", amplitude=0.18)

widgets.HBox([lead, bass, texture], layout=widgets.Layout(gap="16px"))
```
