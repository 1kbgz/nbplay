```python
from nbplay import TransportWidget

transport = TransportWidget(
    bpm=118.0,
    time_signature_num=4,
    time_signature_den=4,
    loop_enabled=True,
    loop_start_bar=0,
    loop_end_bar=2,
)
transport
```
