# 09 MixerWidget Tutorial

Create three channels and set gain/pan on each strip.

```python
from nbplay import MixerWidget

mixer = MixerWidget()
mixer.add_channel("Lead")
mixer.add_channel("Bass")
mixer.add_channel("Drums")

mixer.set_channel_gain(0, 0.78)
mixer.set_channel_pan(0, 0.2)
mixer.set_channel_gain(1, 0.9)
mixer.set_channel_pan(1, -0.15)
mixer.set_channel_gain(2, 0.58)
mixer.master_gain = 0.84

mixer
```
