# 08 Full Session WalkthroughBuild a complete Session with transport links and mixer routing.

```python
import mathimport ipywidgets as widgetsfrom IPython.display import displayfrom nbplay import SamplerWidget, SequencerWidget, Session, SettingsWidget, SynthWidgetsettings = SettingsWidget(sample_rate=44100, channels=2, buffer_size=512)lead = SynthWidget(oscillator_type="saw", frequency=660.0, amplitude=0.55)bass = SynthWidget(oscillator_type="square", frequency=110.0, amplitude=0.65)sampler = SamplerWidget(attack=0.005, decay=0.08, sustain=0.15, release=0.12, max_voices=6)sample_rate = 44100frame_count = int(sample_rate * 0.18)sample_data = [    math.sin(2 * math.pi * 880 * index / sample_rate) * (1 - index / frame_count)    for index in range(frame_count)]sampler.load_sample(sample_data, sample_rate=sample_rate, root_note=60, name="Notebook Blip")lead_seq = SequencerWidget(length=8, bpm=118.0)for step_index, note in enumerate([72, 76, 79, 83, 79, 76, 74, 71]):    lead_seq.set_step(step_index, note=note, velocity=102, active=True)bass_seq = SequencerWidget(length=8, bpm=118.0)for step_index, velocity in enumerate([118, 64, 88, 64, 110, 64, 92, 64]):    bass_seq.set_step(step_index, note=48, velocity=velocity, active=True)drum_seq = SequencerWidget(length=8, bpm=118.0)for step_index in [0, 3, 4, 7]:    drum_seq.set_step(step_index, note=60, velocity=112, active=True)session = Session(bpm=118.0, time_signature=(4, 4))session.add_track("Lead", lead_seq, lead)session.add_track("Bass", bass_seq, bass)session.add_track("Drums", drum_seq, sampler)session.mixer.set_channel_gain(0, 0.78)session.mixer.set_channel_pan(0, 0.25)session.mixer.set_channel_gain(1, 0.92)session.mixer.set_channel_pan(1, -0.18)session.mixer.set_channel_gain(2, 0.58)session.mixer.master_gain = 0.86overview = widgets.VBox([    widgets.HBox([settings, session.transport], layout=widgets.Layout(gap="16px")),    widgets.HBox([lead, bass, sampler], layout=widgets.Layout(gap="16px")),    widgets.HBox([lead_seq, bass_seq, drum_seq], layout=widgets.Layout(gap="16px")),    session.mixer,], layout=widgets.Layout(gap="16px"))display(overview)print(session)
```
